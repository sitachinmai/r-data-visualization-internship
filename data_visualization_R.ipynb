```python

%%R

# Load the built-in Iris dataset
data(iris)

# Display first six rows
head(iris)

# Check dataset structure
str(iris)

# Dataset summary
summary(iris)
```

```python

%%R

barplot(table(iris$Species),
        main = "Flower Species Distribution",
        xlab = "Species",
        ylab = "Number of Flowers",
        col = "skyblue")
```

```python

%%R

plot(iris$Sepal.Length, iris$Petal.Length,
     main = "Sepal Length vs Petal Length",
     xlab = "Sepal Length (cm)",
     ylab = "Petal Length (cm)",
     col = as.numeric(iris$Species),
     pch = 19)

legend("topleft",
       legend = levels(iris$Species),
       col = 1:3,
       pch = 19)
```

```python

%%R

hist(iris$Petal.Length,
     main = "Distribution of Petal Length",
     xlab = "Petal Length (cm)",
     ylab = "Frequency",
     col = "lightgreen",
     border = "white")
```

```python

%%R

avg_petal <- aggregate(Petal.Length ~ Species,
                       data = iris,
                       FUN = mean)

plot(1:3, avg_petal$Petal.Length,
     type = "b",
     pch = 19,
     xaxt = "n",
     main = "Average Petal Length by Species",
     xlab = "Species",
     ylab = "Average Petal Length (cm)",
     col = "blue")

axis(1, at = 1:3, labels = avg_petal$Species)
```
